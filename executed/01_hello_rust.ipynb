## Variables and arithmetic
Rust infers types for most local values.

```rust
let a = 6;
let b = 7;
println!("{} × {} = {}", a, b, a * b);
```

## A simple data collection

```rust
let measurements = vec![1.0_f64, 2.0, 3.0, 4.0];
let mean = measurements.iter().sum::<f64>() / measurements.len() as f64;
println!("Mean: {mean}");
```
