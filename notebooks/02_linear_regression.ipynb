# 02 — Linear regression in Rust
An AI/ML research mini-experiment in a Jupyter notebook, using **ordinary least squares** without extra crates. We'll learn the line `y = 2x + 1` from data and predict `y` when `x = 6`.

```rust
let x = vec![0.0_f64, 1.0, 2.0, 3.0, 4.0];
let y = vec![1.0_f64, 3.0, 5.0, 7.0, 9.0];
println!("Observations: {}", x.len());
```

## Compute the means

```rust
let mean_x = x.iter().sum::<f64>() / x.len() as f64;
let mean_y = y.iter().sum::<f64>() / y.len() as f64;
println!("mean x = {mean_x}, mean y = {mean_y}");
```

## Fit the model
The slope is covariance divided by variance; the intercept ensures the fitted line passes through the mean point.

```rust
let numerator: f64 = x.iter().zip(y.iter()).map(|(a, b)| (a - mean_x) * (b - mean_y)).sum();
let denominator: f64 = x.iter().map(|a| (a - mean_x).powi(2)).sum();
assert!(denominator > 0.0, "x values must vary");
let slope = numerator / denominator;
let intercept = mean_y - slope * mean_x;
println!("slope = {slope}, intercept = {intercept}");
```

## Make a prediction

```rust
let input = 6.0;
let prediction = slope * input + intercept;
assert!((prediction - 13.0).abs() < 1e-10);
println!("Prediction at x={input}: {prediction}");
```
